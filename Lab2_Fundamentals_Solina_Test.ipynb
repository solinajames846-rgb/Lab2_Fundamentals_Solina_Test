```python
import random 
SURNAME = "Solina"
STUDENT_ID = "TUPM-26-2302"
SEED_NUM = int("".join(filter(str.isdigit, STUDENT_ID.split('-')[-1])))
print(f"--- Ntebook Initialized ---")
print(f"Surname: {SURNAME}")
print(f"Student ID: {STUDENT_ID}")
print(f"Derived Seed Number: {SEED_NUM}")
```

```python
# 1. Generate sensor data using surname and SEED_NUM
random.seed(SEED_NUM + len(SURNAME))

# Mix valid numbers, out-of-range values, and unparseable text entries
mock_pool = ["23.5", "45.1", "98.2", "12.0", "INVALID_DATA", "72.4", "ERROR_404", "55.0", "-10.5", "115.3"]
generated_readings = [random.choice(mock_pool) for _ in range(8)]

# Predefined operating ranges
LOW_MAX = 30.0
NORMAL_MAX = 70.0

valid_readings = []
invalid_readings = []
classifications = {"Low": 0, "Normal": 0, "High": 0}
execution_log = []

# 2. Process and validate entries
for idx, entry in enumerate(generated_readings):
    try:
        val = float(entry)
        if val < 0 or val > 100:  # Physical sensor boundaries limit check
            raise ValueError("Value out of physical sensor bounds")
        
        valid_readings.append(val)
        
        # 3. Classify valid readings
        if val <= LOW_MAX:
            category = "Low"
        elif val <= NORMAL_MAX:
            category = "Normal"
        else:
            category = "High"
            
        classifications[category] += 1
        execution_log.append(f"Index {idx}: '{entry}' parsed successfully as {category}.")
        
    except ValueError:
        invalid_readings.append(entry)
        execution_log.append(f"Index {idx}: '{entry}' rejected as Invalid Input.")

# 4. Display Assessment Data & Summary
print("================ ASSESSMENT DATA ================")
print(f"Generated Sensor Data: {generated_readings}")
print(f"Valid Readings:        {valid_readings}")
print(f"Invalid Entries:       {invalid_readings}")
print(f"\nClassification Results: {dict(classifications)}")
print("\nExecution Log:")
for log in execution_log:
    print(f"  [LOG] {log}")

print("\n================ FINAL OUTPUT ================")
print(f"Total Processed: {len(generated_readings)} | Valid: {len(valid_readings)} | Invalid: {len(invalid_readings)}")
print(f"Status Summary: Low ({classifications['Low']}), Normal ({classifications['Normal']}), High ({classifications['High']})")
```

```python
# 1. Generate unique signal using surname and SEED_NUM
random.seed(SEED_NUM * len(SURNAME))
char_pool = SURNAME + str(SEED_NUM) + " @#$_!"
generated_signal = "".join(random.choice(char_pool) for _ in range(15))

# 2. Process and normalize signal (e.g., uppercase transformation for standard report processing)
normalized_signal = generated_signal.upper()

# 3. Analyze character-by-character
analysis_report = []
counts = {"Letters": 0, "Digits": 0, "Spaces": 0, "Special Chars": 0}

for idx, char in enumerate(normalized_signal):
    if char.isalpha():
        char_type = "Letter"
        counts["Letters"] += 1
    elif char.isdigit():
        char_type = "Digit"
        counts["Digits"] += 1
    elif char.isspace():
        char_type = "Space"
        counts["Spaces"] += 1
    else:
        char_type = "Special Char"
        counts["Special Chars"] += 1
    
    analysis_report.append(f"Pos {idx}: '{char}' -> {char_type}")

# Determine overall signal health profile classification
if counts["Special Chars"] > (len(normalized_signal) * 0.3):
    classification = "Noisy/Corrupted Signal"
else:
    classification = "Stable Data Stream"

# 4. Display Assessment Data & Diagnostic Report
print("================ ASSESSMENT DATA ================")
print(f"Generated Signal: {generated_signal}")
print(f"Processed Signal: {normalized_signal}")
print("\nCharacter Analysis:")
for item in analysis_report:
    print(f"  {item}")

print(f"\nSignal Classification: {classification}")
print(f"Execution Log: Analyzed {len(normalized_signal)} frames sequentially.")

print("\n================ FINAL OUTPUT ================")
print(f"Diagnostic Breakdown: {counts}")
print(f"System Operational Status: {classification}")
```

```python
# 1. Generate unique password credentials and limit bounds using surname and seed
random.seed(SEED_NUM - len(SURNAME))
secure_suffix = str(random.randint(100, 999))
generated_password = f"{SURNAME[:3].upper()}_{secure_suffix}" # E.g., SOL_234
attempt_limit = random.randint(3, 5) # Adaptive limit between 3 to 5 based on key metrics

# Simulated user attempt sequences (simulating real-world inputs checking bounds)
simulated_attempts = [
    f"{SURNAME[:3].upper()}_000", 
    "WRONG_PASS", 
    generated_password, # Correct password placed here for test simulation
    "LATE_TRY"
]

attempts_made = 0
access_result = "DENIED"
final_system_state = "LOCKED"
execution_log = []

# 2, 3. Process, normalize, and verify sequential matching
for attempt in simulated_attempts:
    if attempts_made >= attempt_limit:
        execution_log.append(f"Attempt block triggered: Max limit reached.")
        break
        
    attempts_made += 1
    execution_log.append(f"Processing attempt #{attempts_made}: Checking candidate '{attempt}'")
    
    # Analyze alignment character by character match
    if attempt == generated_password:
        access_result = "GRANTED"
        final_system_state = "UNLOCKED"
        execution_log.append("Success: Credential authentication verification match found.")
        break
    else:
        execution_log.append("Fail: String character sequence mismatch detected.")

if final_system_state == "LOCKED":
    execution_log.append("Security Protocol Triggered: System permanently locked down.")

# 4. Display Assessment Data
print("================ ASSESSMENT DATA ================")
print(f"Generated Password: {generated_password}")
print(f"Attempt Limit:      {attempt_limit}")
print(f"Attempts Made:      {attempts_made}")
print(f"Access Result:      {access_result}")
print(f"Final System State: {final_system_state}")
print("\nExecution Log:")
for log in execution_log:
    print(f"  [AUTH] {log}")

print("\n================ FINAL OUTPUT ================")
print(f"Terminal Authentication Status: Access {access_result} | Core Unit is {final_system_state}")
```
